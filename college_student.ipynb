```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
df= pd.read_csv('college student.csv')
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Outstate"], bins=30, kde=True)

plt.title("Distribution of Out-of-State Tuition")
plt.xlabel("Tuition Fee")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Grad.Rate"], bins=20)

plt.title("Graduation Rate Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.countplot(x="Private", data=df)
plt.title("Private vs Public Colleges")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.barplot(x="Private",
            y="Outstate",
            data=df)
plt.title("Average Tuition by College Type")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Enroll"],
             bins=25)
plt.title("Enrollment Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["S.F.Ratio"])
plt.title("Student Faculty Ratio")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Top10perc",
                y="Grad.Rate",
                data=df)
plt.title("Top 10% Students vs Graduation Rate")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Expend",
                y="Grad.Rate",
                hue="Private",
                data=df)
plt.title("Educational Expenditure vs Graduation Rate")
plt.show()
```

```python
sns.pairplot(df[[
    "Apps",
    "Accept",
    "Enroll",
    "Outstate",
    "Expend",
    "Grad.Rate"
]])
plt.show()
```

```python
top_grad = df.sort_values(
    by="Grad.Rate",
    ascending=False
).head(10)
top_grad
```

```python
top_expend = df.sort_values(
    by="Expend",
    ascending=False
).head(10)
top_expend
```
